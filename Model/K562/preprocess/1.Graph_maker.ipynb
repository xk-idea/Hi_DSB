```python
import os
import dgl
import numpy as np
import torch

NUM = [24926, 49246 - 24926, 69049 - 49246, 88165 - 69049, 106257 - 88165, 123369 - 106257, 139283 - 123369,
       153920 - 139283, 168042 - 153920,
       181596 - 168042, 195097 - 181596, 208483 - 195097, 220000 - 208483, 230735 - 220000, 240989 - 230735,
       250025 - 240989, 258145 - 250025,
       265953 - 258145, 271866 - 265953, 278169 - 271866, 282982 - 278169, 288113 - 282982, 303641 - 288113,
       309579 - 303641]

Threshold = 0

for i in range(1, 23):
       hic = np.loadtxt('HiC/chr' + str(i) + ".txt")


       index = hic[:, 2] >= Threshold
       hic = hic[index]

       node_1 = np.hstack((hic[:, 0], hic[:, 1])) - 1
       node_2 = np.hstack((hic[:, 1], hic[:, 0])) - 1
       edata  = np.hstack((hic[:, 2], hic[:, 2]))

       # graph maker
       u, v = torch.tensor(node_1).int(), torch.tensor(node_2).int()
       g = dgl.graph((u, v), num_nodes=NUM[i-1])
       #edge_index = torch.tensor(edge_list, dtype=torch.long).t().contiguous()


       # g.edata["edge_feature"] = torch.tensor(normalized_edata)
       g.edata["edge_feature"] = torch.tensor(edata)

       # label maker
       res = []
       with open('Label/chr' + str(i) + '_label.txt','r') as file:
           lines = file.readlines()
           for aa in range(len(lines)):
               line = lines[aa].strip().split('\t')
               res.append(int(line[0]))
    
       y = torch.tensor(res)
       y = torch.squeeze(y)
       g.ndata["label"] = y

       # id maker
       # g.ndata['node_id'] = g.nodes().reshape(g.nodes().shape[0], 1)
       # g.ndata['node_id'] = torch.arange(NUM[i-1]).reshape(NUM[i-1], 1)

       # # deg maker
       # g.ndata['deg'] = g.out_degrees().float().clamp(min=1)

       # save dgls
       dgl.save_graphs("DGL/chr" + str(i) + ".dgl", g)
```
